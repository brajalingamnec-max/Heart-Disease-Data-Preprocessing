```python
# Create a backup before cleaning

df_before_imputation = df.copy()

# Identify columns containing missing values

missing_columns = df.columns[df.isnull().any()]

print("Columns with Missing Values:")
print(missing_columns)
```

```python
# Replace missing values using mode

for column in missing_columns:

    mode_value = df[column].mode()[0]

    df[column] = df[column].fillna(mode_value)

    print(column, "replaced with mode:", mode_value)
```

```python
outlier_report = {}

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_report[column] = len(outliers)

    print("\nAttribute:", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Number of Outliers:", len(outliers))
```

```python
outlier_summary = []

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = (
        (df[column] < lower) |
        (df[column] > upper)
    ).sum()

    outlier_summary.append({
        'Attribute': column,
        'Lower Limit': lower,
        'Upper Limit': upper,
        'Outlier Count': count
    })

outlier_df = pd.DataFrame(outlier_summary)

print(outlier_df)
```

```python
for column in numerical_columns:

    plt.figure(figsize=(8, 4))

    sns.boxplot(x=df[column])

    plt.title("Outlier Detection - " + column)

    plt.xlabel(column)

    plt.show()
```

```python
for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print("\nOutliers in:", column)

    print(outliers[[column, 'target']])
```

```python
df_capped = df.copy()

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df_capped[column] = df[column].clip(
        lower=lower,
        upper=upper
    )

print("IQR capping demonstration completed!")
```

```python
comparison = pd.DataFrame({
    'Original Mean': df[numerical_columns].mean(),
    'Capped Mean': df_capped[numerical_columns].mean(),
    'Original Std': df[numerical_columns].std(),
    'Capped Std': df_capped[numerical_columns].std()
})

print(comparison)
```

```python
X = df_encoded.drop('target', axis=1)

y = df_encoded['target']

print("Input Features Shape:", X.shape)

print("Target Variable Shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X = df.drop('target', axis=1)
y = df['target']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ('numerical', StandardScaler(), numerical_columns),
        ('categorical', OneHotEncoder(handle_unknown='ignore'), categorical_columns)
    ]
)

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform testing data using the same fitted preprocessing
X_test_processed = preprocessor.transform(X_test)

print("Preprocessing Completed!")

print("Training Shape:", X_train_processed.shape)
print("Testing Shape:", X_test_processed.shape)
```

```python
# Extract the standardized numerical features
scaled_features = X_train_processed[:, :len(numerical_columns)]

print("Standardized Numerical Feature Means:")
print(np.asarray(scaled_features.mean(axis=0)).ravel())

print("\nStandardized Numerical Feature Standard Deviations:")
print(np.asarray(scaled_features.std(axis=0)).ravel())
```

```python
from sklearn.preprocessing import MinMaxScaler

minmax_scaler = MinMaxScaler()

normalized_train = minmax_scaler.fit_transform(
    X_train[numerical_columns]
)

normalized_test = minmax_scaler.transform(
    X_test[numerical_columns]
)

print("Min-Max Normalization Completed!")

print(pd.DataFrame(
    normalized_train,
    columns=numerical_columns
).head())
```
