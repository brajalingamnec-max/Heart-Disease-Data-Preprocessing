```python
# Extract upper triangle of correlation matrix

upper_triangle = correlation_matrix.where(
    np.triu(
        np.ones(correlation_matrix.shape),
        k=1
    ).astype(bool)
)

# Identify pairs with absolute correlation above 0.5

high_correlations = (
    upper_triangle.stack()
    .reset_index()
)

high_correlations.columns = [
    'Attribute 1',
    'Attribute 2',
    'Correlation'
]

high_correlations = high_correlations[
    high_correlations['Correlation'].abs() > 0.5
]

display(high_correlations)
```

```python
anomaly_columns = [
    'alcohol',
    'volatile acidity',
    'sulphates'
]

anomaly_summary = []

for column in anomaly_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = (
        (df[column] < lower) |
        (df[column] > upper)
    ).sum()

    anomaly_summary.append({
        'Attribute': column,
        'Lower Limit': lower,
        'Upper Limit': upper,
        'Potential Anomalies': count
    })

anomaly_df = pd.DataFrame(anomaly_summary)

display(anomaly_df)
```

```python
fig, axes = plt.subplots(
    1, 3,
    figsize=(15, 5)
)

for ax, column in zip(axes, anomaly_columns):

    sns.boxplot(
        data=df,
        x=column,
        ax=ax
    )

    ax.set_title("Boxplot of " + column)

plt.suptitle("Potential Anomaly Detection")
plt.tight_layout()
plt.show()
```

```python

# Create a copy of the dataset

transformed_df = df.copy()

print("Original Dataset Shape:", df.shape)
print("Transformed Dataset Shape:", transformed_df.shape)

display(transformed_df.head())
```

```python

# Create Alcohol Level

transformed_df['alcohol_level'] = pd.cut(
    transformed_df['alcohol'],
    bins=[0, 9, 11, 20],
    labels=['Low', 'Medium', 'High']
)

# Create Acidity Level

transformed_df['acidity_level'] = pd.cut(
    transformed_df['volatile acidity'],
    bins=[0, 0.4, 0.7, 2],
    labels=['Low', 'Medium', 'High']
)

print("New Features Created Successfully")

display(
    transformed_df[
        ['alcohol', 'alcohol_level',
         'volatile acidity', 'acidity_level']
    ].head(10)
)
```

```python

print("Alcohol Level Distribution:")
display(transformed_df['alcohol_level'].value_counts())

print("Acidity Level Distribution:")
display(transformed_df['acidity_level'].value_counts())
```

```python

# Data Aggregation

aggregation_result = transformed_df.groupby('quality').agg({
    'alcohol': ['mean', 'min', 'max'],
    'volatile acidity': ['mean', 'min', 'max'],
    'citric acid': ['mean', 'max'],
    'sulphates': ['mean', 'max'],
    'density': ['mean', 'min', 'max']
})

print("Data Aggregation Results:")
display(aggregation_result)
```

```python

# Group-wise analysis

group_analysis = transformed_df.groupby(
    ['alcohol_level', 'acidity_level'],
    observed=True
).agg(
    Average_Quality=('quality', 'mean'),
    Average_Alcohol=('alcohol', 'mean'),
    Average_Acidity=('volatile acidity', 'mean'),
    Wine_Count=('quality', 'count')
).reset_index()

print("Group-wise Analysis:")
display(group_analysis)
```
