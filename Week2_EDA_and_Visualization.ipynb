```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_style("whitegrid")

# Display all columns
pd.set_option('display.max_columns', None)

print("All libraries imported successfully!")
```

```python
# Download the dataset

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"

df = pd.read_csv(url, sep=';')

print("Dataset downloaded successfully!")
```

```python
# Display first five records

df.head()
```

```python
# Display last five records

df.tail()
```

```python
# Check dataset dimensions

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Identify numerical columns

numerical_columns = df.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

print("Numerical Attributes:")
print(numerical_columns)

print("\nTotal Numerical Attributes:", len(numerical_columns))
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df['alcohol'], bins=20, edgecolor='black')

plt.title("Distribution of Alcohol Content")
plt.xlabel("Alcohol Percentage")
plt.ylabel("Frequency")

plt.show()
```

```python
df[numerical_columns].hist(
    figsize=(16, 12),
    bins=20,
    edgecolor='black'
)

plt.suptitle(
    "Histograms of All Numerical Attributes",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x='alcohol',
    kde=True
)

plt.title("Distribution of Alcohol Content")
plt.xlabel("Alcohol Percentage")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x=df['alcohol'])

plt.title("Boxplot of Alcohol Content")
plt.xlabel("Alcohol Percentage")

plt.show()
```

```python
plt.figure(figsize=(15, 8))

sns.boxplot(data=df[numerical_columns])

plt.title("Boxplots of All Numerical Attributes")
plt.xticks(rotation=45)

plt.show()
```

```python
skewness = df[numerical_columns].skew()

print("Skewness of Numerical Attributes:")
print(skewness)
```

```python
univariate_summary = pd.DataFrame({
    'Mean': df[numerical_columns].mean(),
    'Median': df[numerical_columns].median(),
    'Standard Deviation': df[numerical_columns].std(),
    'Skewness': df[numerical_columns].skew(),
    'Minimum': df[numerical_columns].min(),
    'Maximum': df[numerical_columns].max()
})

display(univariate_summary)
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='alcohol',
    y='quality'
)

plt.title("Alcohol Content vs Wine Quality")
plt.xlabel("Alcohol Percentage")
plt.ylabel("Wine Quality")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='volatile acidity',
    y='quality'
)

plt.title("Volatile Acidity vs Wine Quality")
plt.xlabel("Volatile Acidity")
plt.ylabel("Wine Quality")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='sulphates',
    y='quality'
)

plt.title("Sulphates vs Wine Quality")
plt.xlabel("Sulphates")
plt.ylabel("Wine Quality")

plt.show()
```

```python
average_alcohol = df.groupby('quality')['alcohol'].mean()

print(average_alcohol)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=average_alcohol.index,
    y=average_alcohol.values
)

plt.title("Average Alcohol Content by Wine Quality")
plt.xlabel("Wine Quality")
plt.ylabel("Average Alcohol Percentage")

plt.show()
```

```python
selected_columns = [
    'alcohol',
    'volatile acidity',
    'sulphates',
    'quality'
]

sns.pairplot(
    df[selected_columns],
    diag_kind='hist'
)

plt.show()
```

```python
selected_columns = [
    'alcohol',
    'volatile acidity',
    'sulphates',
    'quality'
]

sns.pairplot(
    df[selected_columns],
    diag_kind='hist'
)

plt.show()
```

```python
correlation = df[
    ['alcohol', 'volatile acidity', 'sulphates', 'quality']
].corr(method='pearson')

print("Correlation Matrix:")
display(correlation)
```
